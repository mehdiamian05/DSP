#### **Fourier and its inverse**

#### This code takes FFT from a signal, then takes inverse FFT, and compare the results visually.

```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# 1. Create a sample signal
# -----------------------------
fs = 1000  # sampling frequency in Hz
duration = 1.0  # seconds

t = np.arange(0, duration, 1/fs)

# Signal = 50 Hz sine wave + 120 Hz sine wave
x = (
    np.sin(2 * np.pi * 50 * t)
    + 0.5 * np.sin(2 * np.pi * 120 * t)
)

# -----------------------------
# 2. Compute FFT
# -----------------------------
X = np.fft.fft(x)

# Frequency axis
freqs = np.fft.fftfreq(len(x), d=1/fs)

# FFT magnitude
magnitude = np.abs(X)

# -----------------------------
# 3. Compute inverse FFT
# -----------------------------
x_reconstructed = np.fft.ifft(X)

# Because of numerical precision, the result may be complex with tiny imaginary parts
x_reconstructed = np.real(x_reconstructed)

# -----------------------------
# 4. Plot original signal
# -----------------------------
plt.figure(figsize=(10, 4))
plt.plot(t, x)
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.title("Original Signal")
plt.grid(True)
plt.show()

# -----------------------------
# 5. Plot FFT magnitude
# -----------------------------
plt.figure(figsize=(10, 4))
plt.plot(freqs, magnitude)
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude")
plt.title("FFT Magnitude Spectrum")
plt.grid(True)
plt.xlim(-200, 200)
plt.show()

# -----------------------------
# 6. Plot reconstructed signal
# -----------------------------
plt.figure(figsize=(10, 4))
plt.plot(t, x_reconstructed)
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.title("Reconstructed Signal Using Inverse FFT")
plt.grid(True)
plt.show()
```
